```python
from bs4 import BeautifulSoup
import requests

url = "http://www.pcm.gov.lb/arabic/listingbydate.aspx?pageid=16"
result = requests.get(url)
doc = BeautifulSoup(result.text, "html.parser")

# Extraction de tous les liens de pagination (numéros de pages)
pagination_links = doc.find_all('a', class_="paging")

# Affichage du contenu des pages
for paragraph in doc.find_all('div', class_="summaryNewsContainer"):
    print(paragraph.text.strip())

# Recherche du lien vers la page suivante
next_page = None
for link in pagination_links:
    # On cherche un lien dont le texte est le numéro de la page suivante
    page_number = link.text.strip()
    try:
        # Si le texte du lien est un nombre et correspond à la page suivante
        if int(page_number) == 5:  # Ici 17 représente la page suivante, ajustez en fonction de votre logique
            next_page = link.get('href="javascript:WebForm_DoPostBackWithOptions(new WebForm_PostBackOptions(&quot;ctl00$maincont$dlPages$ctl02$lbtnPage&quot;, &quot;&quot;, true, &quot;&quot;, &quot;&quot;, false, true)"')  # Récupération du href de la page suivante
            break
    except ValueError:
        # Si le lien ne contient pas de numéro de page (par exemple "Suivant"), ignorer
        continue

if next_page:
    # Construire l'URL complète pour la page suivante
    base_url = "http://www.pcm.gov.lb/arabic/listingbydate.aspx?pageid=16"
    next_page_url = base_url + next_page
    print("URL de la page suivante:", next_page_url)
else:
    print("Pas de page suivante trouvée.")
```

```python
from selenium import webdriver
from bs4 import BeautifulSoup
import time

# Initialisation du navigateur Web (ici, Chrome en mode "headless")
options = webdriver.ChromeOptions()
options.add_argument("--headless")  # Si vous ne voulez pas ouvrir de fenêtre de navigateur
driver = webdriver.Chrome(options=options)

# URL de la première page
url = "http://www.pcm.gov.lb/arabic/listingbydate.aspx?pageid=16"
driver.get(url)

# Boucle pour parcourir les pages suivantes
while True:
    # Attendre que la page soit complètement chargée
    time.sleep(2)

    # Extraire le HTML de la page actuelle
    soup = BeautifulSoup(driver.page_source, 'html.parser')

    # Extraire et afficher les articles ou informations de la page actuelle
    paragraphs = soup.find_all('div', class_="summaryNewsContainer")
    for paragraph in paragraphs:
        print(paragraph.text.strip())

    # Chercher le lien pour la page suivante (en fonction de son texte ou de sa classe)
    next_page = soup.find('a', class_="paging", text="2")  # Vous pouvez ajuster le texte ou la classe

    # Si un lien pour la page suivante est trouvé
    if next_page:
        # Trouver le bouton pour cliquer dessus (basé sur le texte ou la classe)
        next_page_button = driver.find_element_by_link_text(next_page.text)
        next_page_button.click()
    else:
        print("Fin des pages.")
        break  # Si aucune page suivante n'est trouvée, arrêter la boucle

# Fermer le navigateur
driver.quit()
```
