```python
# convertir 'titre' en liste de tokens
df["tokens"] = df["titre"].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) else [])

print("\nExemple de tokens :")
print(df["tokens"].iloc[0])

# Convertir le label 'yes'/'no' en 1/0
df["label"] = df["label"].map({"yes": 1, "no": 0})

```
